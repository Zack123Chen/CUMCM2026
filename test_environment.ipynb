```python
data = {
    "student": ["A", "B", "C", "D", "E"],
    "math": [85, 92, 78, 96, 88],
    "physics": [82, 90, 75, 94, 91]
}

df = pd.DataFrame(data)

df
```

```python
df.describe()
```

```python
df["average"] = (df["math"] + df["physics"]) / 2

df
```

```python
df_sorted = df.sort_values(
    by = "average",
    ascending = False
)

df_sorted
```

```python
import matplotlib.pyplot as plt

plt.bar(df["student"] , df["average"])

plt.xlabel("Student")
plt.ylabel("Average Score")
plt.title("Average Score of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df["student"], df["average"])

plt.xlabel("Student")
plt.ylabel("Average")
plt.title("Average Score of Student")

plt.savefig(
    "figures/test_plot.png",
    dpi = 300,
    bbox_inches = "tight"
)

plt.show()
```
