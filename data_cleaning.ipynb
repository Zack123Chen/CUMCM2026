```python
import pandas as pd
import numpy as np

print("pandas:", pd.__version__)
print("numpy:", np.__version__)
```

```python
df = pd.DataFrame({
    "name": ["A", "B", "C", "C", "D"],
    "age": [20, 21, 22, 22, 200],
    "score": [88, 92, 76, 76, 95],
    "date": [
        "2026-08-01",
        "2026-08-02",
        "2026-08-03",
        "2026-08-03",
        "2026-08-05"
    ],
    "height_cm": [170, 175, 168, 168, 180]
})

# 再故意制造一个缺失值
df.loc[1, "score"] = np.nan

df
```

```python
print("数据规模：", df.shape)

print("\n列名：")
print(df.columns)

print("\n数据类型：")
print(df.dtypes)

print("\n前5行：")
print(df.head())

print("\n基本信息：")
df.info()

print("\n数据统计：")
df.describe()
```

```python
print("每列缺失值统计:\n",df.isna().sum())

print("每行重复值统计：\n", df.duplicated().sum())

```

```python
df_clean = df.drop_duplicates()

df_clean
```

```python
df_clean = df_clean.copy()

score_median = df_clean["score"].median()
print(score_median)

df_clean["score"] = df_clean["score"].fillna(score_median)

df_clean
```

当一个 DataFrame 是从另一个 DataFrame 筛选出来时，Pandas 有时不知道你修改的是：原数据的一部分，还是一个独立副本。加上copy()可以避免歧义

```python
df_clean["age"].between(0,120)

df_clean.loc[~df_clean["age"].between(0,120), "age"] = np.nan

df_clean

age_median = df_clean["age"].median()
df_clean["age"] = df_clean["age"].fillna(age_median)

df_clean
```

```python
print(df_clean)

print("\n缺失值：")
print(df_clean.isna().sum())

print("\n描述统计：")
print(df_clean.describe())
```

```python
# 将字符串转化为实际的日期
df_clean.dtypes

df_clean["date"] = pd.to_datetime(df_clean["date"])

df_clean.dtypes

df_clean["day"] = df_clean["date"].dt.day

df_clean
```

```python
# 数据筛选
df_clean[df_clean["score"] > 85]
```

```python
# groupby
df_clean["group"] = ["A", "A", "B", "B"]

df_clean.groupby(by="group", as_index=False)[["score", "age"]].mean()

df_clean.groupby("group")["score"].agg(["mean", "max", "min", "median"])
```

```python
# merge：把两张表拼起来
df_extra = pd.DataFrame({
    "name": ["A", "B", "C", "D"],
    "city": ["Beijing", "Shanghai", "Harbin", "Hangzhou"]
})

df_extra

df_merged = pd.merge(left=df_clean, right=df_extra, on="name",how="left")

df_merged
```

```python
# pivot_table：做透视表
df_clean["gender"] = ["M", "F", "M", "F"]

df_clean.pivot_table(values="score", index="group", columns="gender", aggfunc="mean")
```

```python
# 练习
df_clean["group"] = ["A", "A", "B", "B"]
# 1. date 转 datetime
df_clean["date"] = pd.to_datetime(df_clean["date"])
df_clean.dtypes

# 2. 筛选 score >= 85 的记录
df_clean[df_clean["score"] >= 85]

# 3. 按 group 求 score 平均值
df_clean.groupby(by="group", as_index=True)["score"].mean()

# 4. 按 group 求 score 的 mean/max/min
df_clean.groupby(by="group", as_index=True)["score"].agg(["mean", "max", "min"])

# 5. 创建 df_extra，并按 name merge
df_extra = pd.DataFrame({
    "name": ["A", "B", "C", "D"],
    "city": ["Beijing", "Shanghai", "Harbin", "Hangzhou"]
})
df_merged = pd.merge(df_clean, df_extra, how="left", on="name")

# 6. 用 pivot_table 求每组平均 score
df_merged.pivot_table(values="score", index="group", aggfunc="mean")
```
