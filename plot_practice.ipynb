```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8,5))
```

fig 指定的是画布的大小，ax 制定的是坐标轴的大小。

```python
x = [1, 2, 3, 4, 5]
y = [2, 5, 4, 8, 7]

fig, ax = plt.subplots()

ax.scatter(x, y)

plt.show()
```

利用 scatter 绘制散点图

```python
x = [1, 2, 3, 4, 5]
y = [2, 5, 4, 8, 7]

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(x, y)

ax.set_title("Scatter Plot")
ax.set_xlabel("X")
ax.set_ylabel("Y")

fig.savefig("figures/scatter.png", dpi=300, bbox_inches="tight") # bbox_inches=tight:按实际图形内容把边缘裁紧一点。

plt.show()
```

设置标题名称和坐标轴名称
保存代码及其参数设置:dpi,保存路径
figsize设置的时候单位是英寸，dpi是清晰度

```python
x = [1, 2, 3, 4, 5]
y = [2, 5, 4, 8, 7]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(x, y, marker="o")

ax.set_title("Line Plot")
ax.set_xlabel("X")
ax.set_ylabel("Y")

plt.show()
```

```python
categories = ["A", "B", "C", "D"]
values = [12, 18, 9, 15]

fig, ax = plt.subplots(figsize=(8, 5))

ax.bar(categories, values)

ax.set_title("Bar Plot")
ax.set_xlabel("Categories")
ax.set_ylabel("Values")

plt.show()
```

柱状图

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame(
    {
        "temperature": [20, 22, 25, 27, 30, 32],
        "sales": [100, 120, 145, 160, 190, 210],
        "humidity": [80, 75, 70, 65, 60, 55],
        "cost": [50, 55, 60, 68, 75, 82],
    }
)

corr = df.corr()

fig, ax = plt.subplots(figsize=(7, 6))

im = ax.imshow(corr, vmin=-1, vmax=1, cmap="coolwarm") # 把一个二维矩阵直接映射成不同颜色

fig.colorbar(im, ax=ax)  # 右边添加颜色标尺

ax.set_xticks(range(len(corr.columns)))
ax.set_yticks(range(len(corr.columns)))

ax.set_xticklabels(corr.columns) #给x,y轴设置标签
ax.set_yticklabels(corr.columns)

for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        ax.text(i, j, f"{corr.iloc[i, j]:.2f}", ha="center", va="center") # 遍历矩阵的每一个格子，把对应数字写进去
        
plt.savefig("figures/correlation_heatmap.png", dpi=300, bbox_inches="tight")

plt.show()
```
